```python
import pandas as pd
df = pd.read_csv("../data/features.csv")
print(df.shape)
display(df.head())

swindon_names = pd.read_excel("../data/Swindon LSOA local names.xlsx")
df_swindon = df[df["LSOA21CD"].isin(swindon_names["LSOA21CD"])].copy()
print(df_swindon.shape)
print(df_swindon["MSOA21CD"].nunique())
```

```python
msoa_check = (df_swindon.groupby(["MSOA21CD", "MSOA21NM"]).size().reset_index(name="Number of LSOAs"))
display(msoa_check)
print("Number of MSOAs:", len(msoa_check))
print("Total number of LSOAs:", msoa_check["Number of LSOAs"].sum())
```

```python
for col in df_swindon.columns:
    print(col)
```

```python
sum_columns = [
    "LSOA GVA Estimates (millions)",
    "Mid-2024 population",
    "Working_Age_Pop",
    "Total_Pop_Mid2024",
    "part_time_employees",
    "full_time_employees",
    "total_employees",
    "bus_stop_count",
    "area_km2"
]

print(df_swindon[sum_columns].dtypes)
print("\nMissing values:")
print(df_swindon[sum_columns].isnull().sum())
```

```python
gva_missing = df_swindon[
    df_swindon["LSOA GVA Estimates (millions)"].isnull()
][
    ["LSOA21CD", "LSOA21NM", "MSOA21CD", "MSOA21NM"]
]

display(gva_missing)
print("\nNumber of MSOAs:", gva_missing["MSOA21CD"].nunique())
```

```python
gva_missing_by_msoa = (gva_missing.groupby(["MSOA21CD", "MSOA21NM"]).size().reset_index(name="Missing GVA LSOAs"))
display(gva_missing_by_msoa)
```

```python
msoa_features = (
    df_swindon
    .groupby(
        ["MSOA21CD", "MSOA21NM"],
        as_index=False
    )
    .agg({
        "LSOA GVA Estimates (millions)": "sum",
        "Mid-2024 population": "sum",
        "Working_Age_Pop": "sum",
        "Total_Pop_Mid2024": "sum",
        "part_time_employees": "sum",
        "full_time_employees": "sum",
        "total_employees": "sum",
        "bus_stop_count": "sum",
        "area_km2": "sum"
    })
)

missing_gva_msoa = df_swindon[
    df_swindon["LSOA GVA Estimates (millions)"].isna()
]["MSOA21CD"].unique()

msoa_features.loc[
    msoa_features["MSOA21CD"].isin(missing_gva_msoa),
    "LSOA GVA Estimates (millions)"
] = np.nan

display(msoa_features)
```

```python
print("MSOA rows:", len(msoa_features))
print("Unique MSOAs:",msoa_features["MSOA21CD"].nunique())
print("Missing MSOA GVA:",msoa_features["LSOA GVA Estimates (millions)"].isnull().sum())
```

```python
check_columns = [
    "Mid-2024 population",
    "Working_Age_Pop",
    "Total_Pop_Mid2024",
    "part_time_employees",
    "full_time_employees",
    "total_employees",
    "bus_stop_count",
    "area_km2"
]

for col in check_columns:
    before = df_swindon[col].sum()
    after = msoa_features[col].sum()

    print(col)
    print("Before:", before)
    print("After:", after)
    print("Same:", round(before, 6) == round(after, 6))
```

```python
print((df_swindon["Mid-2024 population"]== df_swindon["Total_Pop_Mid2024"]).all())
```

```python
msoa_columns = [
    "LU_large_2025_msoa",
    "LU_medium_2025_msoa",
    "LU_micro_2025_msoa",
    "LU_small_2025_msoa",
    "LU_total_2025_msoa",
    "LU_pct_micro_2025_msoa",
    "LU_pct_small_2025_msoa",
    "LU_pct_medium_2025_msoa",
    "LU_pct_large_2025_msoa"
]

for col in msoa_columns:
    max_values = df_swindon.groupby("MSOA21CD")[col].nunique().max()
    print(col, max_values)
```

```python
msoa_business = (df_swindon.groupby(["MSOA21CD", "MSOA21NM"], as_index=False)[msoa_columns].first())
display(msoa_business.head())
```

```python
msoa_features = msoa_features.merge(msoa_business, on=["MSOA21CD", "MSOA21NM"], how="left")
print(msoa_features.shape)
display(msoa_features.head())
```

```python
other_msoa_columns = [
    "LU_diversity_1-HHI_2025",
    "pct_enterprise_0_to_49_thousand_2025_msoa",
    "pct_enterprise_100_to_249_thousand_2025_msoa",
    "pct_enterprise_1000_to_1999_thousand_2025_msoa",
    "pct_enterprise_10000_to_49999_thousand_2025_msoa",
    "pct_enterprise_2000_to_4999_thousand_2025_msoa",
    "pct_enterprise_250_to_499_thousand_2025_msoa",
    "pct_enterprise_50_to_99_thousand_2025_msoa",
    "pct_enterprise_500_to_999_thousand_2025_msoa",
    "pct_enterprise_5000_to_9999_thousand_2025_msoa",
    "pct_enterprise_50000plus_thousand_2025_msoa",
    "turnover_diversity_1-HHI_2025_msoa",
    "total_enterprises_2025_msoa",
    "share_enterprises_creative_2025_msoa",
    "share_enterprises_foundational_2025_msoa",
    "share_enterprises_green_aligned_2025_msoa",
    "share_enterprises_industrial_2025_msoa",
    "share_enterprises_kibs_2025_msoa",
    "share_enterprises_anchor_2025_msoa",
#    "enterprises_per_1k_residents_2025",
    "share_enterprises_public_2025_msoa",
    "share_enterprises_private_2025_msoa"
]

for col in other_msoa_columns:
    max_values = df_swindon.groupby("MSOA21CD")[col].nunique().max()
    print(col, max_values)
```

```python
other_msoa_data = (df_swindon.groupby(["MSOA21CD", "MSOA21NM"], as_index=False)[other_msoa_columns].first())
msoa_features = msoa_features.merge(other_msoa_data, on=["MSOA21CD", "MSOA21NM"],how="left")

msoa_features["enterprises_per_1k_residents_2025"] = (
    msoa_features["total_enterprises_2025_msoa"]
    / msoa_features["Total_Pop_Mid2024"]
    * 1000
)

print(msoa_features.shape)
display(msoa_features.head())
```

```python
msoa_features["bus_stop_density"] = (msoa_features["bus_stop_count"] / msoa_features["area_km2"])
display(msoa_features[["MSOA21CD", "MSOA21NM", "bus_stop_count", "area_km2", "bus_stop_density"]].head())
```

```python
msoa_features["employment_rate_per_pop"] = (
    msoa_features["total_employees"]
    / msoa_features["Total_Pop_Mid2024"]
)

display(
    msoa_features[
        ["MSOA21CD", "MSOA21NM", "total_employees",
         "Total_Pop_Mid2024", "employment_rate_per_pop"]
    ].head()
)
```

```python
msoa_features = msoa_features.drop(columns=["Mid-2024 population"])
msoa_features = msoa_features.rename(columns={"LSOA GVA Estimates (millions)": "MSOA GVA Estimates (millions)"})

print(msoa_features.shape)
display(msoa_features.head())
```
