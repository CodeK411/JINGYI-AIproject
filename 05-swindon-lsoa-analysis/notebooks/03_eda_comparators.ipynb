```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
```

```python
features_path = "../data/features.csv"
names_path = "../data/Swindon LSOA local names.xlsx"
```

```python
df_features = pd.read_csv(features_path)
df_names = pd.read_excel(names_path)
```

```python
df_merged = pd.merge(
    df_features, 
    df_names[['LSOA21CD', 'Local LSOA21 Name']], 
    on='LSOA21CD', 
    how='left'
)

cols = df_merged.columns.tolist()
cols = cols[:2] + [cols[-1]] + cols[2:-1]
df_merged = df_merged[cols]
```

```python
print(f"Dataset: {df_merged.shape[0]} rows, {df_merged.shape[1]} columns")
display(df_merged.head())
df_merged.info()
```

```python
df_features_clean = df_features.drop_duplicates(subset=['LSOA21CD'], keep='first')

# Swindon codes
swindon_codes = set(df_names['LSOA21CD'])

# All non-Swindon areas
df_non_swindon = df_features_clean[
    ~df_features_clean['LSOA21CD'].isin(swindon_codes)
].copy()

# Give a placeholder name column
df_non_swindon['Local LSOA21 Name'] = df_non_swindon['LSOA21NM']

print(f"Non-Swindon areas: {len(df_non_swindon)}")
display(df_non_swindon.head())

missing_names = df_non_swindon['Local LSOA21 Name'].isnull().sum()
if missing_names == 0:
    print("All matched")
else:
    print(f"Still {missing_names} area missing ")
```

```python
print(len(df_non_swindon))
print(df_non_swindon.shape[1])
display(df_non_swindon.head())
df_non_swindon.info()
```

```python
missing_counts = df_non_swindon.isnull().sum()
missing_report = missing_counts[missing_counts > 0].sort_values(ascending=False)
missing_percentage = (missing_report / len(df_non_swindon)) * 100

if missing_report.empty:
    print("No missing values")
else:
    print(f"{len(missing_report)} features lost")
    missing_df = pd.DataFrame({
        'Missing Values (Count)': missing_report,
        'Percentage (%)': missing_percentage.round(2)
    })
    display(missing_df)
```

```python
sns.set_theme(style="whitegrid", context="talk")
plt.figure(figsize=(14, 10))

x_var = 'LSOA GVA Estimates (millions)'
y_var = 'IMD_decile'

df_non_swindon = df_non_swindon.dropna(subset=[x_var, y_var, 'Local LSOA21 Name']).copy()

scatter = sns.scatterplot(
    data=df_non_swindon, 
    x=x_var, 
    y=y_var, 
    alpha=0.7, 
    edgecolor="k",
    s=150
)

x_median = df_non_swindon[x_var].median()
y_median = df_non_swindon[y_var].median()

plt.axvline(x_median, color='red', linestyle='--', alpha=0.5, label='GVA Median')
plt.axhline(y_median, color='red', linestyle='--', alpha=0.5, label=f'{y_var} Median')

# The smaller the IMD, the poorer the region IMD < 0.10 
# x_var > 0.90 < y_median
outliers = df_non_swindon[
    (df_non_swindon[y_var] < df_non_swindon[y_var].quantile(0.10)) | 
    ((df_non_swindon[x_var] > df_non_swindon[x_var].quantile(0.90)) & (df_non_swindon[y_var] < y_median))
]

for idx, row in outliers.iterrows():
    plt.text(
        row[x_var] + 1,
        row[y_var], 
        row['Local LSOA21 Name'], 
        fontsize=9, 
        color='black',
        weight='bold'
    )

plt.title(f"Non-Swindon Inclusive Growth Matrix: GVA vs {y_var}", fontsize=16, weight='bold', pad=20)
plt.xlabel("Economic Output: LSOA GVA Estimates (Millions £)", fontsize=12, weight='bold')
plt.ylabel(f"Deprivation Index: {y_var} (1=Poor, 10=Rich)", fontsize=12, weight='bold') 
plt.legend()
plt.tight_layout()

plt.show()
```

```python
sns.set_theme(style="whitegrid", context="talk")
plt.figure(figsize=(14, 10))

x_var = 'LSOA GVA Estimates (millions)'
y_var = 'VOA_total_RV_pct_change'

scatter = sns.scatterplot(
    data=df_non_swindon, 
    x=x_var, 
    y=y_var, 
    alpha=0.7, 
    edgecolor="k",
    s=150
)

x_median = df_non_swindon[x_var].median()
y_median = df_non_swindon[y_var].median()
plt.axvline(x_median, color='red', linestyle='--', alpha=0.5, label='GVA Median')
plt.axhline(y_median, color='red', linestyle='--', alpha=0.5, label='RV Growth Median')

# 1. Extremely high growth rate (>90%), output very low (<x_median)
# 2. Extremely high output (>90%), growth rate lags behind the market average (<y_median)
outliers = df_non_swindon[
    ((df_non_swindon[y_var] > df_non_swindon[y_var].quantile(0.90)) & (df_non_swindon[x_var] < x_median)) | 
    ((df_non_swindon[x_var] > df_non_swindon[x_var].quantile(0.90)) & (df_non_swindon[y_var] < y_median))
]

for idx, row in outliers.iterrows():
    plt.text(
        row[x_var] + 1,
        row[y_var], 
        row['Local LSOA21 Name'], 
        fontsize=9, 
        color='black',
        weight='bold'
    )

plt.title("Non-Swindon Economic Matrix: Absolute Wealth vs Dynamic Growth", fontsize=16, weight='bold', pad=20)
plt.xlabel("Economic Output: LSOA GVA Estimates (Millions £)", fontsize=12, weight='bold')
plt.ylabel("Commercial Vitality: RV % Change (2023-2026)", fontsize=12, weight='bold')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
sns.set_theme(style="whitegrid", context="talk")
plt.figure(figsize=(14, 10))

x_var = 'LSOA GVA Estimates (millions)'
y_var = 'employment_rate_per_pop'

scatter = sns.scatterplot(
    data=df_non_swindon, 
    x=x_var, 
    y=y_var, 
    alpha=0.7, 
    edgecolor="k",
    s=150
)

x_median = df_non_swindon[x_var].median()
y_median = df_non_swindon[y_var].median()
plt.axvline(x_median, color='red', linestyle='--', alpha=0.5, label='GVA Median')
plt.axhline(y_median, color='red', linestyle='--', alpha=0.5, label='Employment Median')

# 1. A working-poor community with extremely high employment (>90%) but extremely low output (<x_median)
# 2. An enclave with extremely high output (>90%) but unable to provide sufficient local employment (<y_median)
outliers = df_non_swindon[
    ((df_non_swindon[y_var] > df_non_swindon[y_var].quantile(0.90)) & (df_non_swindon[x_var] < x_median)) | 
    ((df_non_swindon[x_var] > df_non_swindon[x_var].quantile(0.90)) & (df_non_swindon[y_var] < y_median))
]

for idx, row in outliers.iterrows():
    plt.text(
        row[x_var] + 1,
        row[y_var], 
        row['Local LSOA21 Name'], 
        fontsize=9, 
        color='black',
        weight='bold'
    )

plt.title("Non-Swindon Inclusive Matrix: Wealth Generation vs Local Job Creation", fontsize=16, weight='bold', pad=20)
plt.xlabel("Economic Output: LSOA GVA Estimates (Millions £)", fontsize=12, weight='bold')
plt.ylabel("Labour Market: Employment per Population Ratio", fontsize=12, weight='bold')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
x_var = 'LSOA GVA Estimates (millions)'
y_var = 'employment_rate_per_pop' 

# Sensitivity test
df_gva = df_non_swindon.dropna(subset=[x_var, y_var]).copy()
outlier_idx = df_gva[x_var].idxmax()
outlier_name = df_gva.loc[outlier_idx, 'Local LSOA21 Name']
df_without_gva_outlier = df_gva.drop(index=outlier_idx)

sensitivity_rows = []
for method in ['pearson', 'spearman']:
    sensitivity_rows.append({
        'sample': 'All Swindon LSOAs with GVA',
        'method': method,
        'n': len(df_gva),
        f'{x_var} vs {y_var}': df_gva[[x_var, y_var]].corr(method=method).iloc[0, 1]
    })
    sensitivity_rows.append({
        'sample': f'Without largest GVA outlier ({outlier_name})',
        'method': method,
        'n': len(df_without_gva_outlier),
        f'{x_var} vs {y_var}': df_without_gva_outlier[[x_var, y_var]].corr(method=method).iloc[0, 1]
    })

sensitivity_table = pd.DataFrame(sensitivity_rows)
sensitivity_table[f'{x_var} vs {y_var}'] = sensitivity_table[f'{x_var} vs {y_var}'].round(3)
display(sensitivity_table)

fig, axes = plt.subplots(2, 2, figsize=(18, 14))
sns.set_theme(style='whitegrid', context='notebook')

# All Data
# GVA distribution histogram
sns.histplot(df_gva[x_var], bins=30, kde=True, ax=axes[0, 0], color='blue')
axes[0, 0].axvline(df_gva[x_var].median(), color='red', linestyle='--', label='Median')
axes[0, 0].axvline(df_gva[x_var].mean(), color='gray', linestyle=':', label='Mean')
axes[0, 0].set_title('1. GVA Distribution: Raw Scale (ALL DATA)', weight='bold', fontsize=13)
axes[0, 0].set_xlabel('LSOA GVA Estimates (millions)')
axes[0, 0].legend()

# Scatter plot (log)
sns.scatterplot(data=df_gva, x=x_var, y=y_var, ax=axes[0, 1], s=90, alpha=0.75, color='blue', edgecolor='white', linewidth=0.8)

axes[0, 1].set_xscale('log')
axes[0, 1].scatter(df_gva.loc[outlier_idx, x_var], df_gva.loc[outlier_idx, y_var], s=180, color='red', edgecolor='black', zorder=5)
axes[0, 1].annotate(outlier_name, (df_gva.loc[outlier_idx, x_var], df_gva.loc[outlier_idx, y_var]), xytext=(8, 8), textcoords='offset points', fontsize=9, weight='bold')
axes[0, 1].set_title('2. GVA vs Employment Rate: Log X-axis (ALL DATA)', weight='bold', fontsize=13)
axes[0, 1].set_xlabel('LSOA GVA Estimates (millions, Log Scale)')
axes[0, 1].set_ylabel('Employment per Population Ratio')

# Outlier Cut
# GVA distribution histogram
sns.histplot(df_without_gva_outlier[x_var], bins=30, kde=True, ax=axes[1, 0], color='green')
axes[1, 0].axvline(df_without_gva_outlier[x_var].median(), color='red', linestyle='--', label='Median')
axes[1, 0].axvline(df_without_gva_outlier[x_var].mean(), color='gray', linestyle=':', label='Mean')
axes[1, 0].set_title(f'3. GVA Distribution (Without {outlier_name})', weight='bold', fontsize=13)
axes[1, 0].set_xlabel('LSOA GVA Estimates (millions)')
axes[1, 0].legend()

# Scatter plot (log)
sns.scatterplot(data=df_without_gva_outlier, x=x_var, y=y_var, ax=axes[1, 1], s=90, alpha=0.75, color='green', edgecolor='white', linewidth=0.8)
axes[1, 1].set_xscale('log') 
axes[1, 1].axvline(df_without_gva_outlier[x_var].median(), color='gray', linestyle='--', alpha=0.5)
axes[1, 1].axhline(df_without_gva_outlier[y_var].median(), color='gray', linestyle='--', alpha=0.5)
axes[1, 1].set_title(f'4. GVA vs Employment: Log X-axis (Without {outlier_name})', weight='bold', fontsize=13)
axes[1, 1].set_xlabel('LSOA GVA Estimates (millions, Log Scale)')
axes[1, 1].set_ylabel('Employment per Population Ratio')

plt.tight_layout(pad=3.0)
plt.show()

# Scatter plot (Linear)
plt.figure(figsize=(12, 8))
sns.set_theme(style='whitegrid', context='talk')

sns.scatterplot(
    data=df_without_gva_outlier, 
    x=x_var, 
    y=y_var, 
    s=150,
    alpha=0.8, 
    color='purple', 
    edgecolor='white', 
    linewidth=1
)

plt.axvline(df_without_gva_outlier[x_var].median(), color='gray', linestyle='--', alpha=0.6, label='GVA Median')
plt.axhline(df_without_gva_outlier[y_var].median(), color='gray', linestyle='--', alpha=0.6, label='Employment Median')

plt.title("The Non-Swindon Reality: True Economic Scatter Cloud (Outlier Removed)", weight='bold', fontsize=16, pad=20)
plt.xlabel("Economic Output: LSOA GVA Estimates (Millions £, Linear Scale)", weight='bold', fontsize=12)
plt.ylabel("Labour Market: Employment per Population Ratio", weight='bold', fontsize=12)
plt.legend()
sns.despine(trim=True)
plt.tight_layout()

plt.show()
```
